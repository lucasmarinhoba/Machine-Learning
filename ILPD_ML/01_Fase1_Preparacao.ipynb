```python
import kagglehub
import pandas as pd
import numpy as np
import os

# Download do dataset via Kaggle
path = kagglehub.dataset_download("uciml/indian-liver-patient-records")
print("Path to dataset files:", path)
print("Arquivos disponíveis:", os.listdir(path))
```

## 1.5 — Análise Inicial da Variável-Alvo

```python
# Distribuição da variável-alvo
contagem_alvo = df["Dataset"].value_counts().sort_index()
proporcao_alvo = df["Dataset"].value_counts(normalize=True).sort_index() * 100

print("Distribuição da variável-alvo (Dataset):")
print(f"  Classe 1 (doença hepática):     {contagem_alvo[1]:>4d} registros ({proporcao_alvo[1]:.1f}%)")
print(f"  Classe 2 (sem doença hepática):  {contagem_alvo[2]:>4d} registros ({proporcao_alvo[2]:.1f}%)")
print(f"\n  Razão entre classes (1:2): {contagem_alvo[1] / contagem_alvo[2]:.2f} : 1")
print(f"\n⚠️  O dataset é desbalanceado: a classe majoritária (doença) representa ~{proporcao_alvo[1]:.0f}% dos dados.")
```

```python
# Verificação de que a variável-alvo contém apenas os valores esperados (1 e 2)
valores_unicos_alvo = sorted(df["Dataset"].unique())
print(f"Valores únicos em 'Dataset': {valores_unicos_alvo}")

assert valores_unicos_alvo == [1, 2], "A variável-alvo contém valores inesperados!"
print("✓ Variável-alvo contém apenas os valores 1 e 2, conforme esperado.")
```

### Observações relevantes para as próximas fases

1. **Desbalanceamento das classes:** A classe de doença hepática representa ~71% dos dados. Isso deverá ser considerado nas estratégias de treinamento e na escolha das métricas de avaliação.

2. **Desproporção de gênero:** O dataset contém ~76% de pacientes masculinos e ~24% femininos. Isso pode refletir a composição real da amostra ou um viés de coleta.

3. **Valores extremos:** Algumas features (bilirrubina, enzimas hepáticas) apresentam valores máximos muito distantes da mediana. Esses outliers serão analisados em detalhe na EDA.

4. **Escala das features:** As features possuem escalas muito diferentes (ex.: Age 4–90 vs. Alkaline_Phosphotase 63–2110). Modelos sensíveis à escala precisarão de padronização.

5. **Imputação necessária:** Os 4 valores ausentes em `Albumin_and_Globulin_Ratio` deverão ser imputados, utilizando parâmetros calculados somente nos dados de treinamento.

6. **Transformação do alvo:** A variável `Dataset` deverá ser convertida de {1, 2} para {1, 0} na fase de pré-processamento.
